# Environment & Config Check

**Stage 02 - Tooling Setup**

This notebook verifies the project interpreter, local configuration, reusable config helper, and NumPy installation. It never prints API-key values.

## 1. Interpreter and Project Location

```python
from pathlib import Path
import sys


def find_project_root() -> Path:
    """Locate project/ without assuming Jupyter's launch directory."""
    cwd = Path.cwd().resolve()
    for base in (cwd, *cwd.parents):
        for candidate in (base, base / 'project'):
            if (candidate / 'src' / 'config.py').is_file():
                return candidate
    raise FileNotFoundError('Could not locate project/src/config.py')


PROJECT_ROOT = find_project_root()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

print('Python version:', sys.version.split()[0])
print('Interpreter:', sys.executable)
print('Project root:', PROJECT_ROOT)
```

## 2. Load `.env` and Verify Configuration

```python
from src.config import get_data_dir, get_key, load_env

environment_loaded = load_env()
api_key_present = bool(get_key('API_KEY'))
data_dir = get_data_dir()

print('Environment loaded:', environment_loaded)
print('API_KEY present:', api_key_present)
print('DATA_DIR:', data_dir)

assert environment_loaded, 'The local project/.env file was not loaded.'
assert api_key_present, 'API_KEY is missing from the local .env file.'
assert data_dir == (PROJECT_ROOT / 'data').resolve()
```

## 3. NumPy Mini-Demo

```python
import numpy as np

values = np.array([2.0, 4.0, 6.0])
doubled = values * 2
mean_value = values.mean()

print('Array:', values)
print('Doubled:', doubled)
print('Mean:', mean_value)
print('NumPy version:', np.__version__)

assert np.array_equal(doubled, np.array([4.0, 8.0, 12.0]))
assert mean_value == 4.0
```
